## Tablets CSV

```python
tablet='https://webscraper.io/test-sites/e-commerce/allinone/computers/tablets'
response=requests.get(tablet)
soup=BeautifulSoup(response.text,'html.parser')
```

```python
box=soup.find_all('div',class_='col-md-4 col-xl-4 col-lg-4')
print(len(box))
```

```python
box[0]
```

```python
box[0].find('a',class_='title').text.strip()
```

```python
box[0].find('h4',class_='price float-end card-title pull-right').text.strip()
```

```python
d={'name':[],'price':[],'reviews':[],'description':[]}
for i in box:
    d['name'].append(i.find('a',class_='title').text.strip())
    d['price'].append(i.find('h4',class_='price float-end card-title pull-right').text.strip())
    d['reviews'].append(i.find('div',class_='ratings').find('span').text)
    d['description'].append(i.find('p',class_='description card-text').text.strip())
```

```python
box[0].find('div',class_='ratings').text.strip()
```

```python
box[0].find('div',class_='ratings').find('span').text
```

```python
import pandas as pd
df=pd.DataFrame(d)
print(df.head(3))
```

## Titanic movie sublikescript


```python
url='https://subslikescript.com/movie/Titanic-120338'
response=requests.get(url)
result=BeautifulSoup(response.text,'html.parser')
```

```python
print(response.status_code)
```

```python
box=result.find('article',class_='main-article')
print(box.h1.text)
title=box.h1.text
```

```python
script=box.find('div',class_='full-script').text.strip()
```

```python
with open(f"{title}.txt","w",encoding="utf-8") as f:
    f.write(script)
```

```python
url='https://subslikescript.com/movies'
response=requests.get(url)
result=BeautifulSoup(response.text,'html.parser')
box=result.find('article',class_="main-article")
list_link=[]
for i in box.find_all("a"):
    list_link.append("https://subslikescript.com"+i['href'])
```

```python
list_link
```

```python
for i in list_link:
    try:
        url=i
        result=requests.get(url)
        soup=BeautifulSoup(result.text,'html.parser')
        box=soup.find('article',class_='main-article')
        title=box.h1.text
        script=box.find('div',class_="full-script").text.strip()
        with open(f"{title}.txt","w") as f:
            f.write(script)
    except:
        print(title)
```
