```python
url3=f'http://api.openweathermap.org/data/2.5/air_pollution/forecast?lat={lat}&lon={lon}&appid={API_key}'
response=requests.get(url3)
result=response.json()
result['list'][0]['components']
```

```python
len(result['list'])
```

```python
ap={'co':[],'no':[],'no2':[],'o3':[],'so2':[],'pm2_5':[],'pm10':[],'nh3':[]}
for i in range (0,96):
    ap['co'].append(result['list'][i]['components']['co'])
    ap['no'].append(result['list'][i]['components']['no'])
    ap['no2'].append(result['list'][i]['components']['no2'])
    ap['o3'].append(result['list'][i]['components']['o3'])
    ap['so2'].append(result['list'][i]['components']['so2'])
    ap['pm2_5'].append(result['list'][i]['components']['pm2_5'])
    ap['pm10'].append(result['list'][i]['components']['pm10'])
    ap['nh3'].append(result['list'][i]['components']['nh3'])
```

```python
df=pd.DataFrame(ap)
```
