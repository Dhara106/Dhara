```python
import requests
```

```python
url="http://127.0.0.1:8000/api/"
res=requests.get(url)
print(res.status_code)
print(res.json())
```

```python
url2="http://127.0.0.1:8000/api/players/"
res2=requests.get(url2)
print(res2.status_code)
print(res2.json())
```

```python
data={
    "name": "abc",
    "runs": 10,
    "innings": 3
}
res3=requests.post(url2,json=data)
print(res3.status_code)
print(res3.json())
```

```python
data={
    "name": "def",
    "runs": 10,
    "innings": 4
}
url2="http://127.0.0.1:8000/api/players/11/"
res3=requests.put(url2,json=data)
print(res3.status_code)
print(res3.json())
# when you want to change multiple values in one time then use put
```

```python
data={
    "runs": 29
}
url2="http://127.0.0.1:8000/api/players/11/"
res3=requests.patch(url2,json=data)
print(res3.status_code)
print(res3.json())
# patch helps to change only one value
```

```python
url2="http://127.0.0.1:8000/api/players/11/"
res3=requests.delete(url2)
print(res3.status_code)
```
