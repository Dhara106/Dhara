```python
import pandas as pd
from bs4 import BeautifulSoup
import numpy as np
```

```python
with open('demo.html') as f:
    res=f.read()
```

```python
soup=BeautifulSoup(res,'html.parser')
```

```python
d={'Name':[],'ROM':[],'RAM':[],'color':[],'Stars':[],'Reviews':[],'Rear Camera':[],'Front Camera':[],'Old_price':[],'New_price':[],'discount':[],'rating':[],'display(cm)':[],'display(inch)':[]}
```

```python
phone=soup.find_all('div',class_='yKfJKb row')
print(len(phone))
```

```python
for i in phone:
    d['Name'].append(i.find('div',class_='KzDlHZ').text.strip().split('(')[0])
    d['ROM'].append(i.find('div',class_='KzDlHZ').text.strip().split('(')[1].split(',')[1][:-1])
    d['RAM'].append(i.find('ul',class_='G4BRas').find_all('li')[0].text.strip().split('|')[0])
    d['color'].append(i.find('div',class_='KzDlHZ').text.strip().split('(')[1].split(',')[0])
    d['Stars'].append(i.find('div',class_='XQDdHH').text.strip())
    d['Reviews'].append(i.find('span',class_="Wphh3N").text.strip().split('&')[1].split()[0])
    d['Rear Camera'].append(i.find('ul',class_='G4BRas').find_all('li')[2].text.strip().split('|')[0])
    d['Front Camera'].append(i.find('ul',class_='G4BRas').find_all('li')[2].text.strip().split('|')[1].split()[0])
    
    try:
        d['Old_price'].append(i.find('div',class_="Nx9bqj _4b5DiR").text.strip()[3:].replace(",",""))
    except:
        d['Old_price'].append('NaN')
        
    try:
        d['New_price'].append(i.find('div',class_="yRaY8j ZYYwLA").text.strip()[3:].replace(",",""))
    except:
        d['New_price'].append('NaN')
        
    try:
        d['discount'].append(i.find('div',class_='UkUFwK').text.strip()[:3])
    except:
        d['discount'].append('NaN')
        
    d['rating'].append(i.find('span',class_="Wphh3N").text.strip().split('&')[0].split()[0].replace(",",""))
    d['display(cm)'].append(i.find('ul',class_='G4BRas').find_all('li')[1].text.strip().split('(')[0].split()[0])
    d['display(inch)'].append(i.find('ul',class_='G4BRas').find_all('li')[1].text.strip().split('(')[1].split()[0])
```
